```python
from transformers import Qwen2VLForConditionalGeneration, AutoProcessor
from qwen_vl_utils import process_vision_info
import torch

model_id = "Qwen/Qwen2-VL-2B-Instruct"   # 2B variant — fits free Colab T4 (15GB VRAM)

model = Qwen2VLForConditionalGeneration.from_pretrained(
    model_id, torch_dtype=torch.bfloat16, device_map="auto"
)
processor = AutoProcessor.from_pretrained(model_id)

# test image — koi bhi satellite-ish sample URL, abhi ke liye placeholder
test_image_url = "https://upload.wikimedia.org/wikipedia/commons/thumb/1/12/Satellite_image_of_Earth.jpg/640px-Satellite_image_of_Earth.jpg"

messages = [
    {
        "role": "user",
        "content": [
            {"type": "image", "image": test_image_url},
            {"type": "text", "text": "Describe what you see in this image."},
        ],
    }
]

text = processor.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
image_inputs, video_inputs = process_vision_info(messages)
inputs = processor(text=[text], images=image_inputs, videos=video_inputs, padding=True, return_tensors="pt").to(model.device)

generated_ids = model.generate(**inputs, max_new_tokens=128)
generated_ids_trimmed = [out_ids[len(in_ids):] for in_ids, out_ids in zip(inputs.input_ids, generated_ids)]
output_text = processor.batch_decode(generated_ids_trimmed, skip_special_tokens=True, clean_up_tokenization_space=True)

print(output_text[0])
```

```python
from transformers import Qwen2VLForConditionalGeneration, AutoProcessor
from qwen_vl_utils import process_vision_info
from PIL import Image
import torch

model_id = "Qwen/Qwen2-VL-2B-Instruct"

# min_pixels/max_pixels processor ko batata hai kitne visual tokens allow karne hain
min_pixels = 256 * 28 * 28
max_pixels = 640 * 28 * 28   # yeh cap OOM rokega

model = Qwen2VLForConditionalGeneration.from_pretrained(
    model_id, torch_dtype=torch.bfloat16, device_map="auto"
)
processor = AutoProcessor.from_pretrained(model_id, min_pixels=min_pixels, max_pixels=max_pixels)

# extra safety — image ko manually bhi resize kar do
img = Image.open(image_path).convert("RGB")
img.thumbnail((768, 768))   # aspect ratio maintain karte hue resize
img.save("test_image_resized.jpg")

messages = [
    {
        "role": "user",
        "content": [
            {"type": "image", "image": "test_image_resized.jpg"},
            {"type": "text", "text": "Describe what you see in this image."},
        ],
    }
]

text = processor.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
image_inputs, video_inputs = process_vision_info(messages)
inputs = processor(text=[text], images=image_inputs, videos=video_inputs, padding=True, return_tensors="pt").to(model.device)

generated_ids = model.generate(**inputs, max_new_tokens=128)
generated_ids_trimmed = [out_ids[len(in_ids):] for in_ids, out_ids in zip(inputs.input_ids, generated_ids)]
output_text = processor.batch_decode(generated_ids_trimmed, skip_special_tokens=True, clean_up_tokenization_space=True)

print(output_text[0])
```
